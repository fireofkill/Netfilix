```python
import numpy as np 
import pandas as pd 
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
data = pd.read_csv("Netflix TV Shows and Movies.csv")
data.head()
```

```python
data = data.drop("id",axis=1)
data = data.drop("index",axis=1)
data = data.drop("imdb_id",axis=1)
```

```python
data["description"]=data["description"].fillna("")
data["age_certification"]=data["age_certification"].fillna(data["age_certification"].mode()[0])
data["imdb_votes"]=data["imdb_votes"].fillna(data["imdb_votes"].mean())
data.isnull().sum()
```

```python
data.iloc[41]["title"] +" "+ data.iloc[41]["description"]
data_text = data ["title"] +" " +data["description"]
data_text = data_text.str.lower()
data_text
```

```python
model = TfidfVectorizer(stop_words="english")

data_number = model.fit_transform(data_text)
```
